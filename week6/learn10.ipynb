```python
import random

guesses = ""
turns = 10

infile = open("words.txt","r")
lines = infile.readlines()
word = random.choice(lines)

while turns > 0:
    failed = 0
    for char in word:
        if char in guesses:
            print(char,end="")
        else:
            print("_",end="")
            failed += 1

    if failed == 0:
        print("사용자 승리")
        break
    
```
